## CONFIGURACIÓN DE PYTORCH Y DATASETS

Preparación de DataLoaders y definición de arquitecturas de redes neuronales con PyTorch.

```python
# Instalación y configuración de PyTorch
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset
from tqdm.auto import tqdm
from sklearn.metrics import classification_report, f1_score, accuracy_score, confusion_matrix
import warnings
warnings.filterwarnings('ignore')

# Configuración del dispositivo
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("=" * 60)
print("CONFIGURACIÓN DE PYTORCH")
print("=" * 60)
print(f"Device: {device}")
print(f"PyTorch version: {torch.__version__}")
if torch.cuda.is_available():
    print(f"CUDA disponible: {torch.cuda.get_device_name(0)}")
    print(f"CUDA version: {torch.version.cuda}")

# Convertir datos a tensores de PyTorch
X_train_tensor = torch.FloatTensor(X_train_scaled)
y_train_tensor = torch.LongTensor(y_train.values)

X_val_tensor = torch.FloatTensor(X_val_scaled)
y_val_tensor = torch.LongTensor(y_val.values)

X_test_tensor = torch.FloatTensor(X_test_scaled)
y_test_tensor = torch.LongTensor(y_test.values)

# Crear TensorDatasets y DataLoaders
batch_size = 64

train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
val_dataset = TensorDataset(X_val_tensor, y_val_tensor)
test_dataset = TensorDataset(X_test_tensor, y_test_tensor)

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

print("\n" + "=" * 60)
print("DATALOADERS CREADOS")
print("=" * 60)
print(f"Batch size: {batch_size}")
print(f"Train batches: {len(train_loader)}")
print(f"Val batches: {len(val_loader)}")
print(f"Test batches: {len(test_loader)}")
print(f"\nInput features: {X_train_tensor.shape[1]}")
print(f"Output classes: {len(torch.unique(y_train_tensor))}")
```

## MODELO 1: MLP (Multi-Layer Perceptron) CON BATCH NORMALIZATION Y DROPOUT

Arquitectura: red neuronal densa con normalización por lotes, dropout y activaciones ReLU.

```python
# Definición del modelo MLP
class MLPClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dims=[256, 128, 64], num_classes=2, dropout=0.3):
        super(MLPClassifier, self).__init__()
        
        layers = []
        prev_dim = input_dim
        
        for hidden_dim in hidden_dims:
            layers.append(nn.Linear(prev_dim, hidden_dim))
            layers.append(nn.BatchNorm1d(hidden_dim))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout))
            prev_dim = hidden_dim
        
        layers.append(nn.Linear(prev_dim, num_classes))
        
        self.network = nn.Sequential(*layers)
        
    def forward(self, x):
        return self.network(x)

# Instanciar modelo
input_dim = X_train_tensor.shape[1]
num_classes = len(torch.unique(y_train_tensor))

model_mlp = MLPClassifier(input_dim=input_dim, 
                          hidden_dims=[256, 128, 64], 
                          num_classes=num_classes,
                          dropout=0.3).to(device)

print("=" * 60)
print("ARQUITECTURA DEL MODELO MLP")
print("=" * 60)
print(model_mlp)
print(f"\nTotal de parámetros: {sum(p.numel() for p in model_mlp.parameters()):,}")
print(f"Parámetros entrenables: {sum(p.numel() for p in model_mlp.parameters() if p.requires_grad):,}")
```

```python
# Función de entrenamiento con early stopping y guardado del mejor modelo
def train_model(model, train_loader, val_loader, criterion, optimizer, 
                num_epochs=100, patience=15, model_name='model'):
    """
    Entrena un modelo con early stopping, reporta métricas y guarda el mejor modelo.
    
    Args:
        model: Modelo PyTorch
        train_loader: DataLoader de entrenamiento
        val_loader: DataLoader de validación
        criterion: Función de pérdida
        optimizer: Optimizador
        num_epochs: Número máximo de épocas
        patience: Épocas sin mejora antes de detener
        model_name: Nombre base para guardar el modelo
    
    Returns:
        history: Diccionario con métricas por época
    """
    
    history = {
        'train_loss': [], 'val_loss': [],
        'train_acc': [], 'val_acc': [],
        'train_f1': [], 'val_f1': []
    }
    
    best_val_loss = float('inf')
    epochs_no_improve = 0
    best_epoch = 0
    
    print("=" * 60)
    print(f"ENTRENAMIENTO DE {model_name.upper()}")
    print("=" * 60)
    print(f"Épocas máximas: {num_epochs}")
    print(f"Patience: {patience}")
    print(f"Optimizador: {optimizer.__class__.__name__}")
    print(f"Learning rate: {optimizer.param_groups[0]['lr']}")
    print("=" * 60)
    
    for epoch in range(num_epochs):
        # Fase de entrenamiento
        model.train()
        train_loss = 0.0
        train_preds = []
        train_labels = []
        
        train_bar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs} [Train]', leave=False)
        for inputs, labels in train_bar:
            inputs, labels = inputs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            train_loss += loss.item() * inputs.size(0)
            _, preds = torch.max(outputs, 1)
            train_preds.extend(preds.cpu().numpy())
            train_labels.extend(labels.cpu().numpy())
            
            train_bar.set_postfix({'loss': f'{loss.item():.4f}'})
        
        train_loss = train_loss / len(train_loader.dataset)
        train_acc = accuracy_score(train_labels, train_preds)
        train_f1 = f1_score(train_labels, train_preds, average='weighted')
        
        # Fase de validación
        model.eval()
        val_loss = 0.0
        val_preds = []
        val_labels = []
        
        with torch.no_grad():
            val_bar = tqdm(val_loader, desc=f'Epoch {epoch+1}/{num_epochs} [Val]', leave=False)
            for inputs, labels in val_bar:
                inputs, labels = inputs.to(device), labels.to(device)
                
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                
                val_loss += loss.item() * inputs.size(0)
                _, preds = torch.max(outputs, 1)
                val_preds.extend(preds.cpu().numpy())
                val_labels.extend(labels.cpu().numpy())
                
                val_bar.set_postfix({'loss': f'{loss.item():.4f}'})
        
        val_loss = val_loss / len(val_loader.dataset)
        val_acc = accuracy_score(val_labels, val_preds)
        val_f1 = f1_score(val_labels, val_preds, average='weighted')
        
        # Guardar métricas
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['train_acc'].append(train_acc)
        history['val_acc'].append(val_acc)
        history['train_f1'].append(train_f1)
        history['val_f1'].append(val_f1)
        
        # Imprimir métricas de la época
        print(f"Epoch {epoch+1}/{num_epochs} | "
              f"Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | "
              f"Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f} | "
              f"Train F1: {train_f1:.4f} | Val F1: {val_f1:.4f}")
        
        # Early stopping y guardado del mejor modelo
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_epoch = epoch + 1
            epochs_no_improve = 0
            torch.save(model.state_dict(), f'{model_name}_best.pth')
            print(f"  ✓ Mejor modelo guardado (val_loss: {val_loss:.4f})")
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"\n⚠ Early stopping activado en época {epoch+1}")
                print(f"  Mejor val_loss: {best_val_loss:.4f} (época {best_epoch})")
                break
    
    # Cargar el mejor modelo
    model.load_state_dict(torch.load(f'{model_name}_best.pth'))
    print(f"\n✓ Modelo '{model_name}_best.pth' cargado con val_loss = {best_val_loss:.4f}")
    
    return history

print("✓ Función de entrenamiento definida")
```

```python
# Entrenar modelo MLP
criterion = nn.CrossEntropyLoss()
optimizer_mlp = optim.Adam(model_mlp.parameters(), lr=0.001, weight_decay=1e-5)

history_mlp = train_model(
    model=model_mlp,
    train_loader=train_loader,
    val_loader=val_loader,
    criterion=criterion,
    optimizer=optimizer_mlp,
    num_epochs=100,
    patience=15,
    model_name='mlp_classifier'
)
```

```python
# Visualizar curvas de aprendizaje del modelo MLP
def plot_training_curves(history, model_name='Model'):
    """
    Grafica las curvas de pérdida, accuracy y F1-score durante el entrenamiento.
    """
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    epochs = range(1, len(history['train_loss']) + 1)
    
    # Loss
    axes[0].plot(epochs, history['train_loss'], 'b-', label='Train Loss', linewidth=2)
    axes[0].plot(epochs, history['val_loss'], 'r-', label='Val Loss', linewidth=2)
    axes[0].set_title(f'{model_name} - Loss', fontsize=14, fontweight='bold')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Loss')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    # Accuracy
    axes[1].plot(epochs, history['train_acc'], 'b-', label='Train Accuracy', linewidth=2)
    axes[1].plot(epochs, history['val_acc'], 'r-', label='Val Accuracy', linewidth=2)
    axes[1].set_title(f'{model_name} - Accuracy', fontsize=14, fontweight='bold')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Accuracy')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    
    # F1-Score
    axes[2].plot(epochs, history['train_f1'], 'b-', label='Train F1-Score', linewidth=2)
    axes[2].plot(epochs, history['val_f1'], 'r-', label='Val F1-Score', linewidth=2)
    axes[2].set_title(f'{model_name} - F1-Score', fontsize=14, fontweight='bold')
    axes[2].set_xlabel('Epoch')
    axes[2].set_ylabel('F1-Score')
    axes[2].legend()
    axes[2].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Análisis de overfitting/underfitting
    final_train_loss = history['train_loss'][-1]
    final_val_loss = history['val_loss'][-1]
    loss_diff = abs(final_val_loss - final_train_loss)
    
    print("\n" + "=" * 60)
    print("ANÁLISIS DE OVERFITTING/UNDERFITTING")
    print("=" * 60)
    print(f"Loss final train: {final_train_loss:.4f}")
    print(f"Loss final val: {final_val_loss:.4f}")
    print(f"Diferencia: {loss_diff:.4f}")
    
    if loss_diff < 0.05:
        print("✓ Modelo bien ajustado (diferencia mínima entre train y val)")
    elif final_val_loss > final_train_loss + 0.1:
        print("⚠ Posible overfitting (val_loss >> train_loss)")
        print("  Recomendaciones: aumentar dropout, regularización, o más datos")
    elif final_train_loss > 0.5:
        print("⚠ Posible underfitting (train_loss alto)")
        print("  Recomendaciones: modelo más complejo, más épocas, o ajustar learning rate")

plot_training_curves(history_mlp, model_name='MLP Classifier')
```

```python
# Evaluación del modelo MLP en el conjunto de prueba
def evaluate_model(model, test_loader, model_name='Model'):
    """
    Evalúa el modelo en el conjunto de prueba y genera reporte detallado.
    """
    model.eval()
    test_preds = []
    test_labels = []
    
    with torch.no_grad():
        for inputs, labels in tqdm(test_loader, desc=f'Evaluando {model_name}'):
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            test_preds.extend(preds.cpu().numpy())
            test_labels.extend(labels.cpu().numpy())
    
    # Métricas generales
    test_acc = accuracy_score(test_labels, test_preds)
    test_f1 = f1_score(test_labels, test_preds, average='weighted')
    
    print("=" * 60)
    print(f"EVALUACIÓN EN TEST SET - {model_name.upper()}")
    print("=" * 60)
    print(f"Accuracy: {test_acc:.4f}")
    print(f"F1-Score (weighted): {test_f1:.4f}")
    
    # Reporte de clasificación detallado
    print("\n" + "=" * 60)
    print("REPORTE DE CLASIFICACIÓN DETALLADO")
    print("=" * 60)
    target_names = ['Present (0)', 'Absent (1)']
    print(classification_report(test_labels, test_preds, target_names=target_names, digits=4))
    
    # Matriz de confusión
    cm = confusion_matrix(test_labels, test_preds)
    
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=target_names, yticklabels=target_names,
                cbar_kws={'label': 'Count'})
    plt.title(f'Matriz de Confusión - {model_name}', fontsize=14, fontweight='bold')
    plt.ylabel('Etiqueta Real')
    plt.xlabel('Etiqueta Predicha')
    plt.tight_layout()
    plt.show()
    
    return {'accuracy': test_acc, 'f1_score': test_f1, 'predictions': test_preds, 'labels': test_labels}

results_mlp = evaluate_model(model_mlp, test_loader, model_name='MLP Classifier')
```

## MODELO 2: RESNET-STYLE (RESIDUAL NETWORK FOR TABULAR DATA)

Arquitectura con bloques residuales que permiten entrenar redes más profundas evitando el problema del gradiente evanescente.

```python
# Definición del bloque residual
class ResidualBlock(nn.Module):
    def __init__(self, dim, dropout=0.3):
        super(ResidualBlock, self).__init__()
        self.block = nn.Sequential(
            nn.Linear(dim, dim),
            nn.BatchNorm1d(dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(dim, dim),
            nn.BatchNorm1d(dim)
        )
        self.relu = nn.ReLU()
        
    def forward(self, x):
        residual = x
        out = self.block(x)
        out += residual  # Skip connection
        out = self.relu(out)
        return out

# Modelo ResNet para datos tabulares
class ResNetClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim=256, num_blocks=3, num_classes=2, dropout=0.3):
        super(ResNetClassifier, self).__init__()
        
        # Capa de proyección inicial
        self.input_proj = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.BatchNorm1d(hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        
        # Bloques residuales
        self.res_blocks = nn.Sequential(*[
            ResidualBlock(hidden_dim, dropout) for _ in range(num_blocks)
        ])
        
        # Capa de clasificación
        self.classifier = nn.Linear(hidden_dim, num_classes)
        
    def forward(self, x):
        x = self.input_proj(x)
        x = self.res_blocks(x)
        x = self.classifier(x)
        return x

# Instanciar modelo ResNet
model_resnet = ResNetClassifier(
    input_dim=input_dim,
    hidden_dim=256,
    num_blocks=4,
    num_classes=num_classes,
    dropout=0.3
).to(device)

print("=" * 60)
print("ARQUITECTURA DEL MODELO RESNET")
print("=" * 60)
print(model_resnet)
print(f"\nTotal de parámetros: {sum(p.numel() for p in model_resnet.parameters()):,}")
print(f"Parámetros entrenables: {sum(p.numel() for p in model_resnet.parameters() if p.requires_grad):,}")
```

```python
# Entrenar modelo ResNet
optimizer_resnet = optim.Adam(model_resnet.parameters(), lr=0.001, weight_decay=1e-5)

history_resnet = train_model(
    model=model_resnet,
    train_loader=train_loader,
    val_loader=val_loader,
    criterion=criterion,
    optimizer=optimizer_resnet,
    num_epochs=100,
    patience=15,
    model_name='resnet_classifier'
)
```

```python
# Visualizar curvas de aprendizaje del modelo ResNet
plot_training_curves(history_resnet, model_name='ResNet Classifier')
```

```python
# Evaluación del modelo ResNet en el conjunto de prueba
results_resnet = evaluate_model(model_resnet, test_loader, model_name='ResNet Classifier')
```
